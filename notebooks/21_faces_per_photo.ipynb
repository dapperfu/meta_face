# 21 — Faces per photo (distributions)

Deeper face-count analysis: histograms of faces per photo for scrfd and dlib_detect, plus per-directory breakdown.

```python
import _analysis  # noqa: F401

from pathlib import Path

import matplotlib.pyplot as plt
import seaborn as sns

from meta_face.analysis import (
    collect_summaries,
    discover_images,
    face_count_distribution,
    faces_per_photo_stats,
    stats_by_directory,
    to_dataframe,
)

ROOT = Path("/tun/steph_pictures")
WORKERS = 8
```

```python
paths = discover_images(ROOT, recursive=True)
summaries = collect_summaries(paths, workers=WORKERS)
df = to_dataframe(summaries)
faces_per_photo_stats(df, tool="both")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

for ax, tool in zip(axes, ["scrfd", "dlib_detect"]):
    dist = face_count_distribution(df, tool=tool)
    sns.barplot(x=dist.index, y=dist.values, ax=ax)
    ax.set_title(f"Faces per photo ({tool})")
    ax.set_xlabel("face count bucket")
    ax.set_ylabel("photos")

plt.tight_layout()
plt.show()
```
