# 24 — Cluster identity stats

Read existing `face.cluster.*` and `face.cluster_dlib.*` sidecar keys. Requires prior `mf cluster` / `hdbscan` runs.

```python
import _analysis  # noqa: F401

from pathlib import Path

import matplotlib.pyplot as plt
import seaborn as sns

from meta_face.analysis import (
    cluster_collection_stats,
    collect_summaries,
    discover_images,
    stats_by_year,
    to_dataframe,
)

ROOT = Path("/tun/steph_pictures")
WORKERS = 8
```

```python
paths = discover_images(ROOT, recursive=True)
summaries = collect_summaries(paths, workers=WORKERS)
df = to_dataframe(summaries)
cluster_collection_stats(df)
```

```python
yearly = stats_by_year(df)
cluster_yearly = df[df["year"].notna()].groupby("year", as_index=False).agg(
    photos=("media_path", "count"),
    cluster_identities=("cluster_unique", "sum"),
    cluster_noise=("cluster_noise", "sum"),
    cluster_dlib_identities=("cluster_dlib_unique", "sum"),
)
cluster_yearly
```

```python
if not cluster_yearly.empty:
    fig, ax = plt.subplots(figsize=(12, 5))
    sns.barplot(data=cluster_yearly, x="year", y="cluster_identities", ax=ax)
    ax.set_title("ArcFace cluster identities per year")
    ax.tick_params(axis="x", rotation=45)
    plt.tight_layout()
    plt.show()
```
