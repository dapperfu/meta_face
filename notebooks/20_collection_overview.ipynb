# 20 — Collection overview (high-level stats)

One-screen summary of a photo tree: image count, sidecar coverage, and average faces per photo (scrfd + dlib).

Set `ROOT` to a single year (`/tun/steph_pictures/2018`) or the full collection (`/tun/steph_pictures`).

```python
import _analysis  # noqa: F401 — repo path setup

from pathlib import Path

from meta_face.analysis import (
    collect_summaries,
    coverage_report,
    discover_images,
    faces_per_photo_stats,
    to_dataframe,
)

ROOT = Path("/tun/steph_pictures")
WORKERS = 8
```

```python
paths = discover_images(ROOT, recursive=True)
print(f"{len(paths):,} images under {ROOT}")

summaries = collect_summaries(paths, workers=WORKERS)
df = to_dataframe(summaries)
```

```python
coverage_report(df)
```

```python
faces_per_photo_stats(df, tool="both")
```
