# 23 — Coverage gaps

List images missing `.scar` sidecars or missing `scrfd` / `dlib_detect` data. Export paths for `mf scan --force`.

```python
import _analysis  # noqa: F401

from pathlib import Path

from meta_face.analysis import (
    collect_summaries,
    coverage_report,
    discover_images,
    gap_paths,
    to_dataframe,
)

ROOT = Path("/tun/steph_pictures")
WORKERS = 8
OUTPUT_DIR = Path("/tmp/meta_face_gaps")
```

```python
paths = discover_images(ROOT, recursive=True)
summaries = collect_summaries(paths, workers=WORKERS)
df = to_dataframe(summaries)
coverage_report(df)
```

```python
gaps = {
    "missing_sidecar": gap_paths(df, gap="sidecar"),
    "missing_scrfd": gap_paths(df, gap="scrfd"),
    "missing_dlib_detect": gap_paths(df, gap="dlib_detect"),
}

for name, paths_list in gaps.items():
    print(f"{name}: {len(paths_list):,}")
```

```python
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
for name, paths_list in gaps.items():
    out = OUTPUT_DIR / f"{name}.txt"
    out.write_text("\n".join(paths_list))
    print(f"wrote {out} ({len(paths_list):,} paths)")
```
