# 22 — Per-year breakdown (20XX folders)

Aggregate face stats by year folder name (`2006`–`2025`). Parent directory must match `20XX` to appear in the year column.

```python
import _analysis  # noqa: F401

from pathlib import Path

import matplotlib.pyplot as plt
import seaborn as sns

from meta_face.analysis import collect_summaries, discover_images, stats_by_year, to_dataframe

ROOT = Path("/tun/steph_pictures")
WORKERS = 8
```

```python
paths = discover_images(ROOT, recursive=True)
summaries = collect_summaries(paths, workers=WORKERS)
df = to_dataframe(summaries)
yearly = stats_by_year(df)
yearly
```

```python
if yearly.empty:
    print("No 20XX year folders found.")
else:
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    sns.barplot(data=yearly, x="year", y="scrfd_mean", ax=axes[0])
    axes[0].set_title("Avg scrfd faces per photo")
    axes[0].tick_params(axis="x", rotation=45)

    sns.barplot(data=yearly, x="year", y="scrfd_total", ax=axes[1])
    axes[1].set_title("Total scrfd faces")
    axes[1].tick_params(axis="x", rotation=45)

    plt.tight_layout()
    plt.show()
```
